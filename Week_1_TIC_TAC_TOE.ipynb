```python
def print_board(b):
    print(f"\n {b[0]} | {b[1]} | {b[2]} \n---|---|---\n {b[3]} | {b[4]} | {b[5]} \n---|---|---\n {b[6]} | {b[7]} | {b[8]} \n")

def check_win(b, p):
    w = [[0,1,2],[3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]]
    return any(b[c[0]] == b[c[1]] == b[c[2]] == p for c in w)

def tic_tac_toe():
    print("=== Tic-Tac-Toe ===\n")
    n1 = input("Player 1 Name: ").strip() or "Player 1"
    n2 = input("Player 2 Name: ").strip() or "Player 2"

    while True:
        s1 = input(f"{n1}, choose symbol (X or O): ").strip().upper()
        if s1 in ["X", "O"]: break
        print("Invalid choice!")

    s2 = "O" if s1 == "X" else "X"
    names = {s1: n1, s2: n2}
    print(f"\n🔹 {n1}: {s1}\n🔸 {n2}: {s2}\n")

    board = [str(i) for i in range(1, 10)]
    curr = s1
    moves = 0

    while moves < 9:
        print_board(board)
        try:
            choice = int(input(f"{names[curr]} ({curr}), choose (1-9): ")) - 1
        except ValueError:
            print("Invalid number!")
            continue

        if choice < 0 or choice > 8 or board[choice] in ["X", "O"]:
            print("Invalid or taken position!")
            continue

        board[choice] = curr
        moves += 1

        if check_win(board, curr):
            print_board(board)
            print(f" {names[curr]} ({curr}) wins! ")
            return

        curr = s2 if curr == s1 else s1

    print_board(board)
    print("It's a draw! ")

if __name__ == "__main__":
    tic_tac_toe()
```
