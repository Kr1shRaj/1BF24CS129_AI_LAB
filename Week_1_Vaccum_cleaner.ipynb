```python
class VacuumCleaner:
    def __init__(self, room_a, room_b, position):
        self.rooms = {
            "A": room_a,
            "B": room_b
        }
        self.position = position

    def display(self):
        print("\nCurrent State:")
        print("Room A:", self.rooms["A"])
        print("Room B:", self.rooms["B"])
        print("Vacuum Position:", self.position)

    def suck(self):
        if self.rooms[self.position] == "Dirty":
            self.rooms[self.position] = "Clean"
            print("Vacuum cleaned Room", self.position)
        else:
            print("Room", self.position, "is already clean.")

    def move_left(self):
        if self.position == "B":
            self.position = "A"
            print("Vacuum moved to Room A")
        else:
            print("Vacuum is already in Room A.")

    def move_right(self):
        if self.position == "A":
            self.position = "B"
            print("Vacuum moved to Room B")
        else:
            print("Vacuum is already in Room B.")


print("===== VACUUM CLEANER =====")

room_a = input("Enter status of Room A (Clean/Dirty): ").capitalize()
room_b = input("Enter status of Room B (Clean/Dirty): ").capitalize()
position = input("Enter initial vacuum position (A/B): ").upper()

vacuum = VacuumCleaner(room_a, room_b, position)

while True:
    print("\n===== MENU =====")
    print("1. Display current state")
    print("2. Suck / Clean room")
    print("3. Move Left")
    print("4. Move Right")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        vacuum.display()

    elif choice == "2":
        vacuum.suck()

    elif choice == "3":
        vacuum.move_left()

    elif choice == "4":
        vacuum.move_right()

    elif choice == "5":
        print("Exiting Vacuum Cleaner AI...")
        break

    else:
        print("Invalid choice! Please try again.")
```
